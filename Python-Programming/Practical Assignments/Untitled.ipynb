```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
```

```python
a=Node(10)
b=Node(20)
c=Node(30)
a.next=b
b.next=c
print(a.data,a.next.data,a.next.next.data)
```

```python
Head=Node(10)
New=Node(20)
New.next=Head
Head=New
print(Head.data,Head.next.data)
```

```python
head=Node(10)
head.next=Node(20)
new=Node(30)

temp=head
while temp.next:
    temp=temp.next
temp.next=new

print(head.data,head.next.data,head.next.next.data)
```

```python
head=Node(10)
head.next=Node(20)
head.next.next=Node(30)

head=head.next
print(head.data,head.next.data)
```

```python
head=Node(10)
head.next=Node(20)
head.next.next=Node(30)

temp=head
while temp.next.next:
    temp=temp.next
temp.next=None

print(head.data, head.next.data)
```

```python
top=Node(10)
top.next=Node(20)
top.next.next=Node(30)

new=Node(40)
new.next=top
top=new
# print(top.data)
print(top.data,top.next.data,top.next.next.data,top.next.next.next.data)
top=top.next
print(top.data,top.next.data,top.next.next.data)
```
