```python
stack=[]
#push
stack.append(10)
stack.append(20)
stack.append(30)
print(stack)
```

```python
#pop
stack.pop()
print(stack)
```

```python
#Display
print("Stack:",stack)
```

```python
#Reverse String
s=input("Enter string:")
print("Reverse:",s[::-1])
```

```python
#reverse stack using another stack
temp=[]
while stack:
    temp.append(stack.pop())

stack=temp
print("Reversed Stack:",stack)
```

```python
q1=[10,20,30]
q2=[40,50]

#Enqueue
q1.append(40)
print(q1)
```

```python
#Dequeue
if q1:
    q1.pop(0)

#Display
print("Queue:",q1)
```

```python
#Reverse queue using stack
stack=[]
while q1:
    stack.append(q1.pop(0))

while stack:
    q1.append(stack.pop())

print("Reversed:",q1)
```

```python
#Merge two queues
q3=q2+q1
print("Merge:",q3)
```

```python
#Check Identical
if q1==q2:
    print("Identical")
else:
    print("Not Identical")
```
